```python
X=df.drop('target',axis=1)
```

```python
y=df['target']
```

```python
scaler=StandardScaler()
```

```python
X_scaler=scaler.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_scaler,y,test_size=0.2,random_state=50)
```

```python
lr=LogisticRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
lr_pre=lr.predict(X_test)
```

```python
rf=RandomForestClassifier(n_estimators=100,random_state=42)
```

```python
rf.fit(X_train,y_train)
```

```python
rf_pre=rf.predict(X_test)
```

```python
print("lr",accuracy_score(y_test,lr_pre))
```

```python
print("Rf accuracy score ",accuracy_score(y_test,rf_pre))
```

```python
model=Sequential()
model.add(Dense(32,activation='relu',input_shape=(X_train.shape[1],)))
model.add(Dense(16,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
ntg=model.fit(X_train,y_train,validation_split=0.2,epochs=50,batch_size=16)
```

```python
loss,accuracy=model.evaluate(X_test,y_test)
```

```python
print("Test Accuracy:",round(accuracy,2))
```
